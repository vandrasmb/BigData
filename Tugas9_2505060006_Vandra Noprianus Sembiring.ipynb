## B. Kueri Bertingkat (Read)

**Kueri 1:** Seluruh menu kategori "Kopi" dengan `tingkat_kepahitan` di atas 3.

```python
kueri1 = menu.find(
    {"kategori": "Kopi", "tingkat_kepahitan": {"$gt": 3}},
    {"_id": 0}
)
for d in kueri1:
    pprint(d)
```

**Kueri 2:** Seluruh menu dengan `harga` di antara 15000 dan 30000 (batas dianggap inklusif).

```python
kueri2 = menu.find(
    {"harga": {"$gte": 15000, "$lte": 30000}},
    {"_id": 0}
)
for d in kueri2:
    pprint(d)
```

**Kueri 3:** Menu yang tidak memiliki field `pedas_level` sama sekali.

```python
kueri3 = menu.find(
    {"pedas_level": {"$exists": False}},
    {"_id": 0}
)
for d in kueri3:
    pprint(d)
```

## C. Pembaruan Data (Update)

Tambahkan field `status` bernilai `"Tersedia"` ke seluruh dokumen dengan `update_many()`.

```python
hasil_update_many = menu.update_many({}, {"$set": {"status": "Tersedia"}})
print("Dokumen yang diubah:", hasil_update_many.modified_count)
```

Ubah `status` menjadi `"Habis"` untuk 2 menu pilihan dengan `update_one()`.

```python
menu_habis = ["Cold Brew", "Nasi Ayam Geprek"]

for nama in menu_habis:
    r = menu.update_one({"nama_menu": nama}, {"$set": {"status": "Habis"}})
    print(f"{nama} -> dokumen diubah: {r.modified_count}")

print()
for d in menu.find({"status": "Habis"}, {"_id": 0, "nama_menu": 1, "status": 1}):
    pprint(d)
```

## D. Integrasi ke Spark

Data diambil dari collection `menu` dengan proyeksi `{"_id": 0}`, karena `ObjectId` tidak
dikenali oleh Spark. Skema didefinisikan secara eksplisit dengan seluruh field dari semua
dokumen. Semua field selain `nama_menu` dibuat *nullable*, sehingga field yang tidak dimiliki
suatu dokumen (mis. `tingkat_kepahitan` pada menu Makanan) menjadi `null`.

```python
from pyspark.sql import SparkSession, functions as F
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, LongType

spark = (SparkSession.builder
         .appName("Tugas9-KopiNusantara")
         .getOrCreate())

docs = list(menu.find({}, {"_id": 0}))

schema = StructType([
    StructField("nama_menu", StringType(), False),
    StructField("harga", LongType(), True),
    StructField("kategori", StringType(), True),
    StructField("tingkat_kepahitan", IntegerType(), True),
    StructField("pedas_level", IntegerType(), True),
    StructField("stok", IntegerType(), True),
    StructField("rekomendasi_barista", StringType(), True),
    StructField("status", StringType(), True),
])

# Setiap dokumen dipetakan ke urutan kolom skema; field yang hilang menjadi None (null)
rows = [tuple(d.get(f.name) for f in schema.fields) for d in docs]
df = spark.createDataFrame(rows, schema=schema)
```

**1. `printSchema()` hasil konversi**

```python
df.printSchema()
```

**Data hasil konversi**

```python
df.show(truncate=False)
```

**2. Rata-rata `harga` per `kategori` menggunakan `groupBy`**

```python
(df.groupBy("kategori")
   .agg(F.round(F.avg("harga"), 2).alias("rata_rata_harga"))
   .show())
```

## E. Kesimpulan

Tantangan utama saat mengonversi data MongoDB ke Spark DataFrame adalah ketidakseragaman
skema antar dokumen. Di MongoDB, setiap dokumen dalam satu collection boleh memiliki field
yang berbeda. Pada collection `menu`, field `tingkat_kepahitan` hanya ada pada menu Kopi,
`pedas_level` hanya ada pada menu Makanan, dan `stok` serta `rekomendasi_barista` hanya
dimiliki sebagian dokumen. Spark DataFrame justru mewajibkan setiap baris memiliki kolom
yang sama dengan tipe data yang tetap. Jika skema hanya disimpulkan dari satu dokumen,
kolom yang tidak ada di dokumen tersebut akan hilang, sedangkan nilai dengan tipe berbeda
dapat memicu error saat konversi. Tantangan lain adalah field `_id` bertipe `ObjectId`
yang tidak dikenali Spark.

Untuk mengatasinya, saya pertama-tama membuang `_id` melalui proyeksi pada `find()`.
Kedua, saya mendefinisikan skema secara eksplisit dengan `StructType` yang memuat seluruh
field dari semua dokumen, dengan semua kolom selain `nama_menu` bersifat nullable. Ketiga,
setiap baris dibentuk secara eksplisit berdasarkan urutan kolom skema, sehingga field yang
tidak ada otomatis menjadi `null`. Dengan cara ini `printSchema()` menampilkan skema yang
konsisten, dan perhitungan rata-rata harga per kategori tetap dapat dijalankan.
